```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('https://media.githubusercontent.com/media/KhalPrawira/Big-Data_Assignment/refs/heads/main/Dataset/global%20air%20pollution%20dataset.csv')
display(df.head())
```

1. ada berapa kota di negara Indonesia dan apa saja kota-kota yang terdapat di dalam dataset?

```python
num_indonesia_cities = len(indonesia_cities)

display(f"Total kota di negara Indonesia: {num_indonesia_cities}")
```

```python
indonesia_cities_df = df[df['Country'] == 'Indonesia']

indonesia_cities = indonesia_cities_df['City'].unique()

display("Kota di Indonesia:")
display(indonesia_cities)
```

2. Berapa nilai AQI Value untuk setiap kota di Indonesia yang terdata?

```python
aqi_values_indonesia = indonesia_cities_df[['City', 'AQI Value']]

display("AQI Value for each city in Indonesia:")
display(aqi_values_indonesia)
```

3. Berapa nilai rata-rata AQI Value untuk setiap kota di Indonesia tersebut?

```python
average_aqi_per_city = indonesia_cities_df.groupby('City')['AQI Value'].mean()

display("Average AQI Value for each city in Indonesia:")
display(average_aqi_per_city)
```

4. Bagaimana urutan Kota-kota tersebut dari rata-rata AQI Value terendah hingga tertinggi?

```python
sorted_average_aqi = average_aqi_per_city.sort_values(ascending=True)

plt.figure(figsize=(20, 10))
sns.barplot(x=sorted_average_aqi.index, y=sorted_average_aqi.values, hue=sorted_average_aqi.index, palette='viridis', legend=False)
plt.xticks(rotation=90)
plt.xlabel("Kota")
plt.ylabel("Rata-rata AQI Value")
plt.title("Rata-rata AQI Value per Kota di Indonesia (Terendah ke Tertinggi)")
plt.tight_layout()
plt.show()
```

# Analisis

Kota dengan rata-rata AQI terendah dimulai dari sebelah kiri memiliki bar yang paling pendek, menunjukkan rata-rata nilai AQI yang paling rendah. Ini berarti kota-kota tersebut memiliki kualitas udara yang relatif lebih baik dibandingkan dengan kota-kota lain dalam dataset. Sebaliknya, kota-kota di sisi kanan plot memiliki bar yang paling tinggi, menunjukkan rata-rata nilai AQI yang paling tinggi. Ini menandakan kota-kota tersebut memiliki kualitas udara yang relatif lebih buruk atau lebih tercemar.



```python
sorted_average_aqi = average_aqi_per_city.sort_values(ascending=True)

top_10_lowest_aqi = sorted_average_aqi.head(10)
top_10_highest_aqi = sorted_average_aqi.tail(10)

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_lowest_aqi.index, y=top_10_lowest_aqi.values, palette='viridis')
plt.xticks(rotation=45, ha='right')
plt.xlabel("Kota")
plt.ylabel("Rata-rata AQI Value")
plt.title("10 Kota dengan Rata-rata AQI Value Terendah di Indonesia")
plt.tight_layout()
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x=top_10_highest_aqi.index, y=top_10_highest_aqi.values, palette='magma')
plt.xticks(rotation=45, ha='right')
plt.xlabel("Kota")
plt.ylabel("Rata-rata AQI Value")
plt.title("10 Kota dengan Rata-rata AQI Value Tertinggi di Indonesia")
plt.tight_layout()
plt.show()
```

5. Di seluruh kota di Indonesia, apakah ada hubungan atau korelasi antara tingginya polusi debu halus dengan polusi gas industri/kendaraan?

```python
correlation = indonesia_cities_df['PM2.5 AQI Value'].corr(indonesia_cities_df['NO2 AQI Value'])

display(f"Korelasi antara nilai PM2.5 AQI dan nilai NO2 AQI di Indonesia: {correlation}")

plt.figure(figsize=(10, 6))
sns.scatterplot(x='PM2.5 AQI Value', y='NO2 AQI Value', data=indonesia_cities_df)
plt.xlabel("Nilai AQI PM2.5")
plt.ylabel("Nilai AQI NO2")
plt.title("Hubungan antara Nilai AQI PM2.5 dan Nilai AQI NO2 di Indonesia")
plt.grid(True)
plt.show()
```

# Analisis:

Nilai korelasi yang dihitung adalah sekitar 0.738. Ini menunjukkan adanya korelasi positif yang kuat antara nilai AQI PM2.5 dan nilai AQI NO2 di kota-kota di Indonesia. Artinya, secara umum, ketika nilai polusi debu halus (PM2.5) meningkat, nilai polusi gas industri/kendaraan (NO2) juga cenderung meningkat, dan sebaliknya. dapat dilihat untuk titik menunjukkan peningkatan ke atas kanan, ini menunjukkan kota-kota dengan nilai PM2.5 yang rendah juga cenderung memiliki nilai NO2 yang rendah, dan kota-kota dengan nilai PM2.5 yang tinggi juga cenderung memiliki nilai NO2 yang tinggi. Korelasi yang kuat ini menunjukkan bahwa sumber polusi yang menghasilkan PM2.5 dan NO2 seringkali saling terkait. Ini bisa berarti bahwa aktivitas seperti lalu lintas kendaraan bermotor dan proses industri merupakan kontributor signifikan terhadap kedua jenis polusi udara ini di Indonesia.
