```python
import pandas as pd

url = "https://media.githubusercontent.com/media/KhalPrawira/Big-Data_Assignment/refs/heads/main/Dataset/Road%20Accident%20Data.csv"
df = pd.read_csv(url)

display(df.head())
```

```python
unique_local_authorities = df['Local_Authority_(District)'].unique()
print(unique_local_authorities)
```

```python
print("Relevant columns in the existing dataframe:")
print(df[['Weather_Conditions', 'Road_Surface_Conditions', 'Light_Conditions']].head())
```

```python
import requests
from datetime import date, timedelta

API_KEY = "XL23858CQULKT2XR5NHAAE8QT"
BASE_URL = "https://weather.visualcrossing.com/VisualCrossingWebServices/rest/services/timeline/"

locations = ["Hackney,UK", "Camden,UK"]
start_date = date(2023, 1, 1)
end_date = date(2023, 1, 31)

weather_data = []

for location in locations:
    current_date = start_date
    while current_date <= end_date:
        date_str = current_date.strftime("%Y-%m-%d")
        url = f"{BASE_URL}{location}/{date_str}/{date_str}?key={API_KEY}&unitGroup=metric&include=days"

        try:
            response = requests.get(url)
            response.raise_for_status()
            data = response.json()
            weather_data.append({'location': location, 'date': date_str, 'data': data})

        except requests.exceptions.RequestException as e:
            print(f"Error fetching data for {location} on {date_str}: {e}")

        current_date += timedelta(days=1)

print(f"Successfully retrieved weather data for {len(weather_data)} days.")
```

```python
import pandas as pd

processed_weather_data = []

for entry in weather_data:
    location = entry['location'].split(',')[0]
    date_str = entry['date']
    if entry['data'] and 'days' in entry['data'] and entry['data']['days']:
        daily_data = entry['data']['days'][0]

        tempmax = daily_data.get('tempmax')
        tempmin = daily_data.get('tempmin')
        temp = daily_data.get('temp')
        precip = daily_data.get('precip')
        precipprob = daily_data.get('precipprob')
        preciptype = daily_data.get('preciptype')
        snow = daily_data.get('snow')
        snowdepth = daily_data.get('snowdepth')
        windgust = daily_data.get('windgust')
        windspeed = daily_data.get('windspeed')
        winddir = daily_data.get('winddir')
        pressure = daily_data.get('pressure')
        visibility = daily_data.get('visibility')
        cloudcover = daily_data.get('cloudcover')
        solarradiation = daily_data.get('solarradiation')
        solarenergy = daily_data.get('solarenergy')
        uvindex = daily_data.get('uvindex')
        conditions = daily_data.get('conditions')
        description = daily_data.get('description')
        icon = daily_data.get('icon')


        processed_weather_data.append({
            'location': location,
            'date': date_str,
            'tempmax': tempmax,
            'tempmin': tempmin,
            'temp': temp,
            'precip': precip,
            'precipprob': precipprob,
            'preciptype': preciptype,
            'snow': snow,
            'snowdepth': snowdepth,
            'windgust': windgust,
            'windspeed': windspeed,
            'winddir': winddir,
            'pressure': pressure,
            'visibility': visibility,
            'cloudcover': cloudcover,
            'solarradiation': solarradiation,
            'solarenergy': solarenergy,
            'uvindex': uvindex,
            'conditions': conditions,
            'description': description,
            'icon': icon,
        })

weather_df = pd.DataFrame(processed_weather_data)

weather_df['date'] = pd.to_datetime(weather_df['date'])

display(weather_df.head())
```

```python
weather_df.to_csv('processed_weather_data.csv', index=False)

print("Processed weather data exported to 'processed_weather_data.csv'")
```

```python
df_renamed = df.rename(columns={'Accident Date': 'date', 'Local_Authority_(District)': 'location'})

df_renamed['date'] = pd.to_datetime(df_renamed['date'], errors='coerce')
weather_df['date'] = pd.to_datetime(weather_df['date'])

print("Date range of accident data:")
print(f"Min date: {df_renamed['date'].min()}, Max date: {df_renamed['date'].max()}")
print("\nDate range of weather data:")
print(f"Min date: {weather_df['date'].min()}, Max date: {weather_df['date'].max()}")


accident_locations_to_merge = ['Hackney', 'Camden']
df_filtered = df_renamed[df_renamed['location'].isin(accident_locations_to_merge)].copy()

if 'Accident_Index' in df_filtered.columns:
    df_filtered = df_filtered.drop(columns=['Accident_Index'])


merged_df = pd.merge(df_filtered, weather_df, on=['date', 'location'], how='outer')

display(merged_df.head())

print("\nColumns in merged_df and their missing value counts:")
print(merged_df.isnull().sum())
```

```python
# Export the merged dataframe to a CSV file
merged_df.to_csv('merged_accident_weather_data.csv', index=False)

print("Merged data exported to 'merged_accident_weather_data.csv'")
```
