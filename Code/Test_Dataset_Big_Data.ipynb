Car Accident dataset Preview

```python
import pandas as pd

url = "https://media.githubusercontent.com/media/KhalPrawira/Big-Data_Assignment/refs/heads/main/Dataset/Road%20Accident%20Data.csv"
df = pd.read_csv(url)

display(df.head())
```

```python
df['Local_Authority_(District)_cleaned'] = df['Local_Authority_(District)'].astype(str).str.replace(' and ', ' ', regex=False).str.strip()

district_percentages = df['Local_Authority_(District)_cleaned'].value_counts(normalize=True) * 100

display(district_percentages)
```

```python
district_counts = df['Local_Authority_(District)'].value_counts()
display(district_counts)
```

```python
district_counts.to_excel('district_counts.xlsx', sheet_name='District Counts')

print("Data 'district_counts' telah berhasil diekspor ke 'district_counts.xlsx'")
```

```python
num_unique_districts = df['Local_Authority_(District)'].nunique()
print(f"Jumlah tipe (nilai unik) dalam kolom Local_Authority_(District): {num_unique_districts}")
```
