# Synthetic Data Generation

```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("zusmani/pakistans-largest-ecommerce-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
df = pd.read_csv('/root/.cache/kagglehub/datasets/zusmani/pakistans-largest-ecommerce-dataset/versions/2/Pakistan Largest Ecommerce Dataset.csv')
```

```python
df.head()
```

```python
# Duplikasi 10% dari dataset untuk tujuan demonstrasi
duplicate_percentage = 0.10
num_rows_to_duplicate = int(len(df) * duplicate_percentage)

# Ambil subset acak dari df untuk diduplikasi
df_to_duplicate = df.sample(n=num_rows_to_duplicate, random_state=42)

# Gabungkan kembali ke DataFrame asli
df_with_duplicates = pd.concat([df, df_to_duplicate], ignore_index=True)

print(f"Original rows: {len(df)}")
print(f"Rows duplicated: {num_rows_to_duplicate}")
print(f"Total rows after duplicating: {len(df_with_duplicates)}")
```

```python
print("Missing values per column:")
print(df_with_duplicates.isnull().sum())
```

```python
# Simpan DataFrame yang sudah diduplikasi ke file CSV baru
new_csv_path = 'Pakistans Largest E-Commerce Dataset.csv'
df_with_duplicates.to_csv(new_csv_path, index=False)
print(f"Dataset dengan duplikat disimpan ke: {new_csv_path}")
```
